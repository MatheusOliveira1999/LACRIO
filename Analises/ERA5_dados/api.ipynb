```python
import cdsapi
import os

# Diretório de saída
output_dir = "dados_era5"
os.makedirs(output_dir, exist_ok=True)

# Dataset e parâmetros
dataset = "reanalysis-era5-pressure-levels"
anos = [str(ano) for ano in range(1940, 2026)]
meses = [f"{m:02d}" for m in range(1, 13)]

# Template base da requisição
def get_request_template(dias):
    return {
        "product_type": "reanalysis",
        "variable": [
            "geopotential",
            "specific_humidity",
            "specific_rain_water_content",
            "temperature",
            "u_component_of_wind",
            "v_component_of_wind"
        ],
        "day": dias,
        "time": [f"{h:02d}:00" for h in range(24)],
        "pressure_level": ["300", "400", "500"],
        "format": "netcdf",
        "area": [-8.5, -77.9, -9.9, -77],
    }

# Inicializa cliente
client = cdsapi.Client()

# Loop principal
for ano in anos:
    for mes in meses:
        output_file = os.path.join(output_dir, f"era5_{ano}_{mes}.nc")

        if os.path.exists(output_file):
            print(f"Arquivo já existe: {output_file}, pulando download.")
            continue

        print(f"Tentando baixar dados para {ano}-{mes}...")

        # Requisição completa (dia 01 ao 31)
        try:
            request = get_request_template([f"{d:02d}" for d in range(1, 32)])
            request["year"] = ano
            request["month"] = mes

            client.retrieve(dataset, request, output_file)
            print(f"✅ Sucesso: {output_file}")
            continue

        except Exception as e:
            print(f"⚠️ Falha no mês completo: {e}")
            print("➡️ Tentando dividir o mês em duas partes...")

        # Divisão: dias 1–15
        try:
            output_file_half1 = os.path.join(output_dir, f"era5_{ano}_{mes}_part1.nc")
            request_half1 = get_request_template([f"{d:02d}" for d in range(1, 16)])
            request_half1["year"] = ano
            request_half1["month"] = mes

            client.retrieve(dataset, request_half1, output_file_half1)
            print(f"✅ Parte 1 salva: {output_file_half1}")
        except Exception as e1:
            print(f"❌ Falha na Parte 1: {e1}")

        # Divisão: dias 16–31
        try:
            output_file_half2 = os.path.join(output_dir, f"era5_{ano}_{mes}_part2.nc")
            request_half2 = get_request_template([f"{d:02d}" for d in range(16, 32)])
            request_half2["year"] = ano
            request_half2["month"] = mes

            client.retrieve(dataset, request_half2, output_file_half2)
            print(f"✅ Parte 2 salva: {output_file_half2}")
        except Exception as e2:
            print(f"❌ Falha na Parte 2: {e2}")
```
