```python
import cdsapi
import os

output_dir = "dados_era5"
os.makedirs(output_dir, exist_ok=True)

dataset = "reanalysis-era5-pressure-levels"
anos = [
    "1940", "1941", "1942", "1943", "1944", "1945", "1946", "1947", "1948", "1949",
    "1950", "1951", "1952", "1953", "1954", "1955", "1956", "1957", "1958", "1959",
    "1960", "1961", "1962", "1963", "1964", "1965", "1966", "1967", "1968", "1969",
    "1970", "1971", "1972", "1973", "1974", "1975", "1976", "1977", "1978", "1979",
    "1980", "1981", "1982", "1983", "1984", "1985", "1986", "1987", "1988", "1989",
    "1990", "1991", "1992", "1993", "1994", "1995", "1996", "1997", "1998", "1999",
    "2000", "2001", "2002", "2003", "2004", "2005", "2006", "2007", "2008", "2009",
    "2010", "2011", "2012", "2013", "2014", "2015", "2016", "2017", "2018", "2019",
    "2020", "2021", "2022", "2023", "2024", "2025"
]  # reduza conforme necessário para testar
meses = [f"{m:02d}" for m in range(1, 13)]

request_template = {
    "product_type": "reanalysis",
    "variable": [
        "geopotential",
        "specific_humidity",
        "specific_rain_water_content",
        "temperature",
        "u_component_of_wind",
        "v_component_of_wind"
    ],
    "day": [f"{d:02d}" for d in range(1, 32)],
    "time": [f"{h:02d}:00" for h in range(24)],
    "pressure_level": ["300", "400", "500"],
    "format": "netcdf",
    "area": [-8.5, -77.9, -9.9, -77],
}

client = cdsapi.Client()

for ano in anos:
    for mes in meses:
        print(f"Baixando dados para o ano {ano}, mês {mes}")
        request = request_template.copy()
        request["year"] = ano
        request["month"] = mes

        output_file = os.path.join(output_dir, f"era5_{ano}_{mes}.nc")

        client.retrieve(dataset, request, output_file)

        print(f"Arquivo salvo em: {output_file}")
```
